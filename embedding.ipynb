```python
from tensorflow.keras.preprocessing.text import one_hot
```

```python
#sentences
sentences = [
    'The cat sat on the mat',
    'The dog sat on the log',
    'The glass of milk',
    'The cup of tea',
    'I am a student',
    'understand the meaning of words',
    'I am a good developer',]
```

```python
# define vocabulary size
vocab_size = 10000
```

```python
#one hot representation
one_hot_repr = [one_hot(words, vocab_size) for words in sentences]
one_hot_repr    
```

```python
# word embedding representation
from tensorflow.keras.layers import Embedding
#from tensorflow.keras.preprocessing.sequences import pad_sequences
from tensorflow.keras.models import Sequential
from tensorflow.keras.utils import pad_sequences
```

```python
import numpy as np
```

```python
sent_length = 8
# padding sequences to ensure uniform length
embedded_docs = pad_sequences(one_hot_repr, maxlen=sent_length, padding='pre')
embedded_docs
```

```python
#feature representation
dim = 10

model = Sequential()
model.add(Embedding(vocab_size, dim, input_length=sent_length))
model.compile(optimizer='adam', loss='mse', metrics=['accuracy'])
```

```python
model.summary()
```

```python
model.predict(embedded_docs) # this will give you the embedding representation of the input sentences
```
